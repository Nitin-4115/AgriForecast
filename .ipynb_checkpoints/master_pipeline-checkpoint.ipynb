```python
import pandas as pd
import numpy as np
import os
import joblib
import xgboost as xgb
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score, confusion_matrix, precision_recall_curve
import warnings
warnings.filterwarnings('ignore')

# Set global visualization styles
plt.style.use('seaborn-v0_8-whitegrid')
sns.set_context("paper", font_scale=1.2)

# 1. Directory Orchestration (Crucial for Streamlit referencing)
PATHS = {
    'raw': 'data/raw/',
    'processed': 'data/processed/',
    'models': 'models/',
    'figures': 'outputs/figures/'
}
for path in PATHS.values():
    os.makedirs(path, exist_ok=True)
print("✅ Directories verified and ready.")

# 2. Data Ingestion
try:
    df_yield = pd.read_csv(f"{PATHS['raw']}yield.csv")
    df_temp = pd.read_csv(f"{PATHS['raw']}temp.csv")
    df_rain = pd.read_csv(f"{PATHS['raw']}rainfall.csv")
    df_pest = pd.read_csv(f"{PATHS['raw']}pesticides.csv")
    print("✅ Raw datasets loaded successfully.")
except FileNotFoundError as e:
    print(f"❌ Error loading data: {e}. Please ensure CSVs are in data/raw/")

# 3. Standardize Columns
df_rain.columns = df_rain.columns.str.strip()

# Standardize Temperature and Rainfall
for df in [df_temp, df_rain]:
    for col in df.columns:
        if 'temp' in col.lower(): df.rename(columns={col: 'Temperature_C'}, inplace=True)
        if 'rain' in col.lower(): df.rename(columns={col: 'Rainfall_mm'}, inplace=True)
        if col.lower() in ['country', 'area']: df.rename(columns={col: 'Area'}, inplace=True)
        if col.lower() == 'year': df.rename(columns={col: 'Year'}, inplace=True)

# Standardize Yield and Pesticides
for name, df in zip(['Yield', 'Pesticides'], [df_yield, df_pest]):
    if 'Country' in df.columns: df.rename(columns={'Country': 'Area'}, inplace=True)
    df['Area'] = df['Area'].astype(str).str.strip()
    df['Year'] = df['Year'].astype(int)

# Extract correct pesticide and yield values
if 'Value' in df_pest.columns: df_pest.rename(columns={'Value': 'Pesticides_Tonnes'}, inplace=True)
df_pest = df_pest[['Area', 'Year', 'Pesticides_Tonnes']]

for col in df_yield.columns:
    if col.lower() in ['value', 'yield', 'hg/ha_yield']: 
        df_yield.rename(columns={col: 'Yield_hg_ha'}, inplace=True)

# 4. Sequential Merge to build Master Dataset
df_master = pd.merge(df_yield, df_pest, on=['Area', 'Year'], how='inner')
df_master = pd.merge(df_master, df_temp, on=['Area', 'Year'], how='inner')
df_master = pd.merge(df_master, df_rain, on=['Area', 'Year'], how='inner')

# 5. Clean Numeric Data and Drop NaNs
numeric_cols = ['Yield_hg_ha', 'Pesticides_Tonnes', 'Temperature_C', 'Rainfall_mm']
for col in numeric_cols:
    df_master[col] = pd.to_numeric(df_master[col], errors='coerce')

df_master = df_master.dropna()

# Save Master Data
master_path = f"{PATHS['processed']}agriforecast_master.csv"
df_master.to_csv(master_path, index=False)
print(f"✅ Data Fusion Complete. Master Dataset Shape: {df_master.shape}")
print(f"✅ Saved to: {master_path}")
```

```python
# ==========================================
# CELL 2: Feature Engineering & Pre-Modeling Visualizations
# ==========================================
print("Starting Feature Engineering...")

# 1. Calculate Historical Medians for Baseline
baseline_df = df_master.groupby(['Area', 'Item'])['Yield_hg_ha'].median().reset_index()
baseline_df.rename(columns={'Yield_hg_ha': 'median_baseline'}, inplace=True)
baseline_df.to_csv(f"{PATHS['processed']}baselines.csv", index=False)

# 2. Join baseline back to calculate training targets
df_engineered = pd.merge(df_master, baseline_df, on=['Area', 'Item'], how='left')

# Target 1: Yield Delta (Percentage shift from historical median)
df_engineered['Yield_Delta'] = (df_engineered['Yield_hg_ha'] - df_engineered['median_baseline']) / df_engineered['median_baseline']

# Target 2: Is Failure (1 if yield drops by > 15% from median, else 0)
df_engineered['Is_Failure'] = (df_engineered['Yield_Delta'] < -0.15).astype(int)

# 3. Create Engineered Feature
df_engineered['Climate_Stress'] = df_engineered['Temperature_C'] / (df_engineered['Rainfall_mm'] + 1)

# Categorical casting for native XGBoost handling
df_engineered['Area'] = df_engineered['Area'].astype('category')
df_engineered['Item'] = df_engineered['Item'].astype('category')

print("✅ Feature Engineering Complete. Generating Visualizations...")

# ==========================================
# VISUALIZATION ENGINE 
# ==========================================
fig_path = PATHS['figures']

# Helper function to save in both formats and show the plot
def save_and_show(fig_name):
    plt.tight_layout()
    plt.savefig(f"{fig_path}{fig_name}.svg", format='svg', bbox_inches='tight')
    plt.savefig(f"{fig_path}{fig_name}.pdf", format='pdf', bbox_inches='tight')
    plt.show()
    plt.close()

# --- Figure 1: Environmental Correlation Heatmap ---
plt.figure(figsize=(8, 7))
corr_cols = ['Yield_hg_ha', 'Pesticides_Tonnes', 'Temperature_C', 'Rainfall_mm', 'Climate_Stress']
sns.heatmap(df_engineered[corr_cols].corr(), annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title('01. Feature Correlation Matrix', pad=20)
save_and_show("01_correlation_matrix")

# --- Figure 2: Global Historical Yield Trend ---
plt.figure(figsize=(7, 5))
sns.lineplot(data=df_engineered, x='Year', y='Yield_hg_ha', estimator='mean', errorbar=None, color='#2ecc71', linewidth=2.5)
plt.title('02. Global Historical Yield Trend (Mean)', pad=20)
plt.ylabel('Average Yield (hg/ha)')
save_and_show("02_yield_trend_line")

# --- Figure 3: Distribution Density of Global Yields ---
plt.figure(figsize=(7, 5))
sns.histplot(df_engineered['Yield_hg_ha'], bins=50, kde=True, color='#3498db')
plt.title('03. Distribution Density of Global Yields', pad=20)
plt.xlabel('Yield (hg/ha)')
save_and_show("03_yield_histogram")

# --- Figure 4: Top 10 High-Yield Regions ---
top_areas = df_engineered.groupby('Area', observed=True)['Yield_hg_ha'].mean().sort_values(ascending=False).head(10).reset_index()
top_areas['Area'] = top_areas['Area'].astype(str)
plt.figure(figsize=(8, 5))
sns.barplot(data=top_areas, x='Yield_hg_ha', y='Area', palette='viridis')
plt.title('04. Top 10 Highest Average Yield Regions', pad=20)
plt.xlabel('Average Yield (hg/ha)')
plt.ylabel('Country / Region')
save_and_show("04_top_countries_bar")

# --- Figure 5: Yield Variance by Crop Species ---
plt.figure(figsize=(10, 7))
sns.boxplot(data=df_engineered, x='Yield_hg_ha', y='Item', palette='Set2')
plt.xticks(rotation=45, ha='right')
plt.title('05. Yield Variance by Crop Species', pad=20)
save_and_show("05_crop_variance_boxplot")

# --- Figure 6: Global Agricultural Climate Profile ---
joint_fig = sns.jointplot(data=df_engineered, x='Temperature_C', y='Rainfall_mm', kind='hex', color='#e74c3c', height=8)
joint_fig.fig.suptitle('06. Climate Jointplot (Temp vs Rainfall)', y=1.02)
save_and_show("06_climate_jointplot")

# --- Figure 15: Climate Stress Impact on Yield Delta ---
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df_engineered, x='Climate_Stress', y='Yield_Delta', hue='Is_Failure', palette={0: '#bdc3c7', 1: '#e74c3c'}, alpha=0.5)
plt.title('15. Climate Stress vs. Yield Shift (Anomaly Trigger)', pad=20)
plt.axhline(0, color='black', linestyle='--')
plt.axhline(-0.15, color='red', linestyle=':', label='Failure Threshold (-15%)')
plt.legend()
save_and_show("15_climate_stress_impact")

# --- Figure 16: Regional Yield Heatmap (Top 15 Countries) ---
top_15_areas = df_engineered['Area'].value_counts().head(15).index
heatmap_data = df_engineered[df_engineered['Area'].isin(top_15_areas)].pivot_table(
    index='Area', columns='Year', values='Yield_hg_ha', aggfunc='mean'
)
plt.figure(figsize=(12, 8))
sns.heatmap(heatmap_data, cmap='YlGnBu', annot=False)
plt.title('16. Regional Yield Density Over Time', pad=20)
save_and_show("16_regional_yield_heatmap")

print("✅ All Pre-Modeling EDA Visualizations generated, displayed, and saved in SVG & PDF!")
```

```python
# ==========================================
# CELL 3: Train Yield Delta Regressor & Evaluate
# ==========================================
print("Preparing Data for Regressor...")

# 1. Define Features and Target
features_reg = ['Year', 'Pesticides_Tonnes', 'Temperature_C', 'Rainfall_mm', 'Climate_Stress', 'Area', 'Item']
X_reg = df_engineered[features_reg]
y_reg = df_engineered['Yield_Delta']

# 2. Train/Test Split
X_train_r, X_test_r, y_train_r, y_test_r = train_test_split(X_reg, y_reg, test_size=0.2, random_state=42)

# 3. Initialize and Train Model
print("Training XGBoost Regressor (this may take a few seconds)...")
reg_model = xgb.XGBRegressor(
    enable_categorical=True, 
    n_estimators=250, 
    learning_rate=0.08, 
    max_depth=6, 
    random_state=42,
    n_jobs=-1
)
reg_model.fit(X_train_r, y_train_r)

# 4. Save Model for Streamlit App
joblib.dump(reg_model, f"{PATHS['models']}agri_delta_regressor.pkl")
print(f"✅ Regressor successfully saved to: {PATHS['models']}agri_delta_regressor.pkl")

# 5. Predict and Calculate Metrics
y_pred_r = reg_model.predict(X_test_r)
rmse = np.sqrt(mean_squared_error(y_test_r, y_pred_r))
r2 = r2_score(y_test_r, y_pred_r)
print(f"📊 Regressor Performance -> RMSE: {rmse:.4f} | R²: {r2:.4f}")

# ==========================================
# REGRESSOR VISUALIZATIONS
# ==========================================
print("Generating Regressor Evaluation Charts...")

# --- Figure 7: Regressor Feature Importance ---
plt.figure(figsize=(10, 6))
importance = reg_model.feature_importances_
feat_imp = pd.DataFrame({'Feature': features_reg, 'Importance': importance}).sort_values(by='Importance', ascending=False)
sns.barplot(data=feat_imp, x='Importance', y='Feature', palette='magma')
plt.title('07. Regression Factors (Feature Importance)', pad=20)
plt.xlabel('Relative Importance')
plt.ylabel('Feature')
save_and_show("07_regressor_importance")

# --- Figure 8: Prediction Quality Scatter ---
plt.figure(figsize=(8, 8))
sns.scatterplot(x=y_test_r, y=y_pred_r, alpha=0.4, color='#3498db')

# Plot the perfect prediction diagonal line
min_val = min(y_test_r.min(), y_pred_r.min())
max_val = max(y_test_r.max(), y_pred_r.max())
plt.plot([min_val, max_val], [min_val, max_val], 'r--', lw=2, label='Perfect Prediction')

plt.title('08. Regression Quality (Actual vs. Predicted Yield Delta)', pad=20)
plt.xlabel('Actual Yield Delta')
plt.ylabel('Predicted Yield Delta')
plt.legend()
save_and_show("08_prediction_scatter")

# --- [NEW] Figure 12: Residual Distribution ---
plt.figure(figsize=(10, 6))
residuals = y_test_r - y_pred_r
sns.histplot(residuals, bins=50, kde=True, color='#9b59b6')
plt.axvline(0, color='black', linestyle='--', linewidth=2)
plt.title('12. Residual Distribution (Prediction Errors)', pad=20)
plt.xlabel('Prediction Error (Actual - Predicted)')
plt.ylabel('Frequency')
save_and_show("12_residual_distribution")

print("✅ All Regressor charts generated and saved successfully!")
```

```python
# ==========================================
# CELL 4: Train Failure Classifier & Evaluate
# ==========================================
from sklearn.metrics import accuracy_score, roc_auc_score, confusion_matrix, precision_recall_curve

print("Preparing Data for Classifier...")

# 1. Define Features and Target (Climate_Stress intentionally omitted)
features_clf = ['Year', 'Pesticides_Tonnes', 'Temperature_C', 'Rainfall_mm', 'Area', 'Item']
X_clf = df_engineered[features_clf]
y_clf = df_engineered['Is_Failure']

# 2. Train/Test Split (Stratified to maintain failure class balance)
X_train_c, X_test_c, y_train_c, y_test_c = train_test_split(
    X_clf, y_clf, test_size=0.2, random_state=42, stratify=y_clf
)

# 3. Train Failure Classifier
print("Training XGBoost Classifier...")
clf_model = xgb.XGBClassifier(
    enable_categorical=True, 
    n_estimators=200, 
    learning_rate=0.1, 
    max_depth=5, 
    random_state=42,
    eval_metric='logloss',
    n_jobs=-1
)
clf_model.fit(X_train_c, y_train_c)

# 4. Save Classifier Model
joblib.dump(clf_model, f"{PATHS['models']}agri_failure_classifier.pkl")
print(f"✅ Classifier successfully saved to: {PATHS['models']}agri_failure_classifier.pkl")

# 5. Predict and Calculate Metrics
y_pred_c = clf_model.predict(X_test_c)
y_prob_c = clf_model.predict_proba(X_test_c)[:, 1]

acc = accuracy_score(y_test_c, y_pred_c)
roc_auc = roc_auc_score(y_test_c, y_prob_c)
print(f"📊 Classifier Performance -> Accuracy: {acc:.4f} | ROC-AUC: {roc_auc:.4f}")

# ==========================================
# CLASSIFIER VISUALIZATIONS
# ==========================================
print("Generating Classifier Evaluation Charts...")

# --- Figure 9: Confusion Matrix ---
plt.figure(figsize=(6, 6))
cm = confusion_matrix(y_test_c, y_pred_c)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', cbar=False, 
            xticklabels=['Normal', 'Failure'], yticklabels=['Normal', 'Failure'])
plt.title('09. Crop Failure Confusion Matrix', pad=20)
plt.xlabel('Predicted Class')
plt.ylabel('Actual Class')
save_and_show("09_failure_confusion_matrix")

# --- Figure 10: Precision-Recall Curve ---
plt.figure(figsize=(8, 6))
precision, recall, thresholds = precision_recall_curve(y_test_c, y_prob_c)
plt.plot(recall, precision, color='#e74c3c', lw=2)
plt.fill_between(recall, precision, alpha=0.2, color='#e74c3c')
plt.title('10. Precision-Recall Curve (Failure Detection)', pad=20)
plt.xlabel('Recall (Sensitivity)')
plt.ylabel('Precision (Positive Predictive Value)')
save_and_show("10_precision_recall_curve")

# --- Figure 11: Classifier Feature Importance ---
plt.figure(figsize=(10, 6))
importance_clf = clf_model.feature_importances_
feat_imp_clf = pd.DataFrame({'Feature': features_clf, 'Importance': importance_clf}).sort_values(by='Importance', ascending=False)
sns.barplot(data=feat_imp_clf, x='Importance', y='Feature', palette='viridis')
plt.title('11. Classification Factors (Feature Importance)', pad=20)
plt.xlabel('Relative Importance')
plt.ylabel('Feature')
save_and_show("11_classifier_importance")

print("✅ All Classifier charts generated and saved successfully!")
print("🎉 Master Pipeline Complete! You can now launch your Streamlit dashboard.")
```
